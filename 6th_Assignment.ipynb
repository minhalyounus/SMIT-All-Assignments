## Write a program to classify a person’s risk level in a police database.
Take input:
* criminal_record (Yes/No)
* number_of_cases
* travel_history_flagged (Yes/No)
* financial_activity_score
* last_activity_days
#### Rules:
* criminal_record = Yes AND cases > 3 → High Risk
* flagged travel + suspicious financial score → High Risk
* no activity for long time (>365 days) → Monitor
* clean record + normal activity → Low Risk
* mixed signals → Medium Risk

```python
criminal_record = input("crinimal record(yes/no)").lower()
number_of_cases = int(input("Enter number of cases"))
travel_history_flagged = input("Travel History flagged(yes/no)").lower()
financial_acticity_score = input("acticity score(high/low)").lower()
last_activity_days = int(input("Enter number of days of activity(high/low):"))

if criminal_record == "yes" and number_of_cases > 3:
    print("High risk")
elif travel_history_flagged == "yes" and financial_acticity_score == "low":
    print("High risk")
elif last_activity_days > 365:
    print("Monitor")
elif criminal_record == "no" and last_activity_days < 365:
    print("Low risk")
else:
    print("Medium risk")
```
