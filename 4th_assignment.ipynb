## A store offers a 10% discount on a product that costs 1500. Write a Python expression to calculate the discounted price using the appropriate arithmetic operator. Ask the user to enter the product price and discount rate.

```python
base_cost = int(input("Enter the product price"))
discount_rate = int(input("Enter the discount rate"))

discounted_price = (base_cost)-((base_cost*discount_rate)/100)

print(f"The normal cost was {base_cost} and with the discount rate was {discount_rate} and final discounted price is {discounted_price}")
```

## You have some apples. Ask the user to enter the total number of apples they have and the number of apples they give to a friend. Write a Python program to calculate how many apples are left.
Which operator will you use to find the remaining apples?

```python
apples = int(input("Enter the number of apples"))
apples_for_friend = int(input("Enter the number of apples you wish to give to a friend"))
apples_left = apples-apples_for_friend

if apples-apples_for_friend <= 0:
    print("Error! You gave more apples than you have!Enter correct values")
else:
   print(f"The original number of apples was {apples} and given to your friend were {apples_for_friend} so the final number of apples left is {apples_left}")
```

## Ask the user to enter the ages of two friends. Write a Python program to check who is older.
Which comparison operator will you use to find out who is older?

```python
friend_1_age = int(input("Enter the age of the first friend"))
friend_2_age = int(input("Enter the age of the second friend"))

if friend_1_age == friend_2_age:
    print(f"Both friends are of equal age {friend_1_age} and {friend_2_age}. Nice try!")
elif friend_1_age > friend_2_age:
    print(f"Friend 1 is older as he is {friend_1_age} years old and second friend just {friend_2_age}")
elif friend_2_age > friend_1_age:
    print(f"Friend 2 is older as he is {friend_2_age} years old and first friend just {friend_1_age}")
```
