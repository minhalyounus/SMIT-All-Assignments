## Student Result Sheet
### Goal:

    Generate report cards for each student in a class.

### Steps:

    Outer loop → 5 students.
    
    Inner loop → 4 subjects.

    If marks < 33 → print “Failed in Subject”.

    Calculate total and grade per student.

### Example:
    Student 1: [80, 90, 70, 85] → Grade: A  
    Student 2: [50, 32, 45, 40] → Failed in Subject 2

```python
for i in range(5):
    print("Student", i + 1)
    marks_list = []
    failed_subjects = []
    total = 0
    
    for j in range(4):
        marks = float(input("Subject " + str(j + 1) + ": "))
        marks_list.append(marks)
        total = total + marks
        if marks < 33:
            failed_subjects.append(j + 1)
            
    for sub in failed_subjects:
        print("Failed in Subject", sub)
        
    avg = total / 4
    
    if avg >= 80:
        grade = "A"
    elif avg >= 70:
        grade = "B"
    elif avg >= 60:
        grade = "C"
    elif avg >= 50:
        grade = "D"
    else:
        grade = "F"
        
    print("Student", i + 1, ":", marks_list, "→ Grade:", grade)
    print()
```
